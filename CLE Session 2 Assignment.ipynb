```sql
SELECT * FROM taxi LIMIT 5;
```

## 1) FILTERING TRIPS GREATER THAN 10 MILES

```sql
SELECT VendorID, tpep_pickup_datetime, tpep_dropoff_datetime, trip_distance FROM taxi where trip_distance > 10 limit 10;
```

## 2) Find trips where the passenger count is greater than 2 and fare is above $20

```sql
SELECT VendorID,passenger_count,fare_amount,total_amount FROM taxi WHERE passenger_count > 2 AND fare_amount > 20 limit 10;
```

## 3) Find trips paid using credit cards

```sql
SELECT VendorID,payment_type,fare_amount,tip_amount,total_amount FROM taxi WHERE payment_type = 1 LIMIT 10;
```

## 4) Calculate total revenue generated by each vendor

```sql
SELECT VendorID,COUNT(*) AS total_trips,SUM(total_amount) AS total_revenue,AVG(fare_amount) AS average_fare FROM taxi WHERE VendorID IS NOT NULL GROUP BY VendorID;
```

## 5) Find the average trip distance for each passenger count

```sql
SELECT passenger_count,COUNT(*) AS total_trips,AVG(trip_distance) AS avg_distance FROM taxi WHERE passenger_count IS NOT NULL GROUP BY passenger_count ORDER BY passenger_count;
```

## 6) Find the top 10 pickup dates with the highest revenue

```sql
SELECT DATE(TRY_TO_TIMESTAMP(tpep_pickup_datetime)) AS pickup_date, COUNT(*) AS total_trips,SUM(total_amount) AS total_revenue FROM taxi WHERE TRY_TO_TIMESTAMP(tpep_pickup_datetime) IS NOT NULL GROUP BY DATE(TRY_TO_TIMESTAMP(tpep_pickup_datetime)) ORDER BY total_revenue DESC LIMIT 10;
```

## 7) Display the total number of taxi trips and total revenue generated by each vendor for different payment methods. Include the payment method names instead of their numerical codes.

```sql
WITH payment_lookup AS (
    SELECT * FROM VALUES(1, 'Credit Card'),(2, 'Cash'),(3, 'No Charge'),(4, 'Dispute'),(5, 'Unknown'),(6, 'Voided Trip') AS p(payment_type, payment_method)
)
SELECT t.VendorID,p.payment_method, COUNT(*) AS total_trips,SUM(t.total_amount) AS total_revenue FROM taxi t LEFT JOIN payment_lookup p ON CAST(t.payment_type AS INT) = p.payment_type GROUP BY t.VendorID,p.payment_method ORDER BY total_revenue DESC;
```

## 8) Display the name of each taxi vendor along with the total number of trips, average trip distance, and total revenue generated by them.

```sql
WITH vendor_lookup AS (
    SELECT * FROM VALUES(1, 'Creative Mobile Technologies'),(2, 'VeriFone Inc.') AS v(VendorID, vendor_name)
)
SELECT v.vendor_name,COUNT(*) AS total_trips,AVG(t.trip_distance) AS avg_distance,SUM(t.total_amount) AS total_revenue FROM taxi t INNER JOIN vendor_lookup v ON t.VendorID = v.VendorID GROUP BY v.vendor_name;
```

## 9) Identify all pickup dates on which the total revenue generated from taxi trips exceeded $10,000. Display the pickup date, total revenue, and number of trips for each date.

```sql
WITH daily_revenue AS (
    SELECT DATE(TRY_TO_TIMESTAMP(tpep_pickup_datetime)) AS pickup_date,SUM(total_amount) AS revenue,COUNT(*) AS total_trips FROM taxi WHERE TRY_TO_TIMESTAMP(tpep_pickup_datetime) IS NOT NULL GROUP BY DATE(TRY_TO_TIMESTAMP(tpep_pickup_datetime))
)
SELECT pickup_date,revenue,total_trips FROM daily_revenue WHERE revenue > 10000 ORDER BY revenue DESC LIMIT 10;
```

## 10) Identify taxi vendors whose average fare amount is greater than $10. Display their vendor ID, total number of trips, and average fare amount.

```sql
WITH vendor_fares AS (
    SELECT VendorID,COUNT(*) AS total_trips,AVG(fare_amount) AS avg_fare FROM taxi GROUP BY VendorID
)
SELECT VendorID,total_trips,avg_fare FROM vendor_fares WHERE avg_fare > 10 ORDER BY avg_fare DESC;
```

## 11) Rank taxi vendors based on the total revenue generated by each vendor.

```sql
SELECT VendorID,total_revenue,RANK() OVER (ORDER BY total_revenue DESC) AS revenue_rank FROM (SELECT VendorID,SUM(total_amount) AS total_revenue FROM taxi WHERE VendorID IS NOT NULL GROUP BY VendorID) AS vendor_data;
```

## 12) Calculate the cumulative revenue generated by each vendor based on pickup date and time.

```sql
SELECT VendorID,tpep_pickup_datetime,total_amount,SUM(total_amount) OVER (PARTITION BY VendorID ORDER BY tpep_pickup_datetime ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_total FROM taxi WHERE VendorID IS NOT NULL ORDER BY VendorID, tpep_pickup_datetime LIMIT 10;
```

## 13) Find the top 3 highest-fare taxi trips for each vendor

```sql
SELECT * FROM (SELECT VendorID,passenger_count,trip_distance,fare_amount,total_amount,ROW_NUMBER() OVER (PARTITION BY VendorID ORDER BY fare_amount DESC) AS trip_rank FROM taxi WHERE VendorID IS NOT NULL AND fare_amount IS NOT NULL) AS ranked_trips WHERE trip_rank <= 3 ORDER BY VendorID, trip_rank;
```
